```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Dataset load karein
df = pd.read_csv('train-selected-columns.csv')

# 2. Sirf numeric columns filter karein (Age, Fare, Survived, etc.)
numeric_df = df.select_dtypes(include=['float64', 'int64'])

# 3. Correlation matrix calculate karein
corr = numeric_df.corr()
 
# 4. Heatmap plot karein
plt.figure(figsize=(9, 7))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)

# 5. Title aur plot display karein
plt.title('Titanic Dataset - Correlation Heatmap', fontsize=14)
plt.show()
```
