```python
import pandas as pd
data = {
    "Student_ID" : [101,102,102,104,105,106],
    "Student" : ["Asha","Ravi","Priya","John","Anu","Kiran"],
    "Department" : ["CSE","CSE","ECE","CSE","ECE","CSE"],
    "Attendence" : [90,98,67,80,87,89],
    "Internal_marks" : [90,89,68,97,89,78],
    "Final_marks" : [90,89,89,None,90,77]
}
df = pd.DataFrame(data)
print(df)
```

Importing dataframe to CSV

```python
df.to_csv("Students.csv", index = False)
```

```python
df = pd.read_csv("Students.csv")
print("Original DataFrame: ")
print(df)
```

2. Explore Dataset

```python
print("\nFirst Five Records: ")
print(df.head(5))
print("\nLast Five Records: ")
print(df.tail(5))
print("/n")
print("\nShape: ")
print(df.shape)
print("\nInfo: ")
print(df.info())
```

4. Missing Values

```python
print("\nMissing Values: ")
print(df.isnull().sum())
```
