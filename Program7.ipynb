```python
# Simple Linear Regression

import numpy as np
import pandas as pd

data = {
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [20, 30, 40, 50, 60]
}

df = pd.DataFrame(data)

X = df["Hours"].to_numpy()
Y = df["Marks"].to_numpy()

mean_X = np.mean(X)
mean_Y = np.mean(Y)

numerator = np.sum((X - mean_X) * (Y - mean_Y))
denominator = np.sum((X - mean_X) ** 2)

m = numerator / denominator
c = mean_Y - m * mean_X

Y_pred = m * X + c

print("Slope:", m)
print("Intercept:", c)
print("Predicted Marks:", Y_pred)

hours = 6
predicted_marks = m * hours + c

print("Hours Studied:", hours)
print("Predicted Marks:", predicted_marks)
```

```python
# Multiple linear regression


import numpy as np
import pandas as pd

df = pd.DataFrame({
    "area": [1000, 1500, 1800, 2000, 2500, 3000],
    "bedrooms": [2, 3, 3, 4, 4, 5],
    "age": [10, 8, 5, 4, 3, 2],
    "price": [200, 300, 350, 400, 480, 600]
})

X = df[["area", "bedrooms", "age"]].to_numpy(dtype=float)
y = df["price"].to_numpy(dtype=float).reshape(-1, 1)

X_mean = X.mean(axis=0)
X_std = X.std(axis=0)
X_std[X_std == 0] = 1

X_scaled = (X - X_mean) / X_std

X_scaled = np.c_[np.ones((X_scaled.shape[0], 1)), X_scaled]

coefficients = np.linalg.pinv(X_scaled) @ y


y_pred = X_scaled @ coefficients


mse = np.mean((y - y_pred) ** 2)

ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - y.mean()) ** 2)
r2 = 1 - ss_res / ss_tot

print("Intercept:", coefficients[0, 0])
print("Coefficients:", coefficients[1:].flatten())
print("Mean Squared Error:", mse)
print("R-squared:", r2)


new_house = np.array([[2200, 3, 5]], dtype=float)

new_house_scaled = (new_house - X_mean) / X_std
new_house_scaled = np.c_[np.ones((1, 1)), new_house_scaled]

predicted_price = new_house_scaled @ coefficients

print("Predicted price:", predicted_price[0, 0])
```
